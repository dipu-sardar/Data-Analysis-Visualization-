21. Create a bar chart for Genre vs Average Rating.

```python
#bar chart করার আগে আমাকে অবশ্যই অবশ্যই সিলেক্ট করে নিতে হবে pandas দিয়ে 
chart = df.groupby('Genre')['Rating'].mean()
plt.figure(figsize=(5,3))
plt.bar(
    chart.index,
    chart.values,
    color = ['green', 'orange', 'black', 'red', 'yellow', 'blue']
)
plt.title('Genre vs Average Rating')
plt.xlabel('Genre')
plt.ylabel('Average Rating')

plt.tight_layout()
plt.show()
```

22. Create a pie chart for Subscription_Type.

```python
char = df.value_counts('Subscription_Type')
plt.figure(figsize=(4,4))
plt.pie(
    chart,
    labels=chart.index,
    colors=['purple', 'red', 'blue', 'yellow', 'orange', 'green'],
    autopct='%1.1f%%'
)
plt.title('Pie chart for Subscription_Type')
plt.xlabel("pie chart")
plt.show()
```

28. Create a stacked bar chart for Genre and Subscription_Type.

```python
stacked_chart = pd.crosstab(
    df['Genre'],
    df['Subscription_Type']
)

stacked_chart.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)

plt.title('Genre vs Subscription Type')
plt.xlabel('Genre')
plt.ylabel('Count')

plt.show()
```

29. Create a horizontal bar chart for top 10 highest rated movies.

```python
highest_rated = df.sort_values(by='Rating', ascending=False).head(10)

plt.barh(
    highest_rated['Movie_Name'],
    highest_rated['Rating'],
    color = ['red', 'green', 'orange']
)

plt.title('Top 10 Highest Rated Movies')
plt.xlabel('Rating')
plt.ylabel('Movie Name')

plt.show()
```

30. Create a line chart showing monthly watch hours growth.

```python
df['Watch_Date'] = pd.to_datetime(df['Watch_Date'])

monthly_growth = df.groupby(
    df['Watch_Date'].dt.month
)['Watch_Hours'].sum()

plt.plot(
    monthly_growth.index,
    monthly_growth.values,
    marker='o'
)

plt.title('Monthly Watch Hours Growth')
plt.xlabel('Month')
plt.ylabel('Total Watch Hours')

plt.show()
```
