# Import Matplotlib 

```python
import matplotlib.pyplot as plt
```

21. Create a bar chart for Category vs Total Sales.

```python
print(df.columns)

dipu = df.groupby('Category')['Total_Amount'].sum()
plt.bar(dipu.index, dipu.values,color='red')

plt.show()
```

22. Create a pie chart for Payment Methods.

```python
payment = df['Payment_Method'].value_counts()

plt.figure(figsize=(10,3))

plt.pie(payment.values, labels=payment.index, colors=['red', 'green', 'blue', 'purple'], autopct='%1.1f%%')
plt.show()
```

23. Create a line chart for daily sales trend.

```python
print(df.columns)
```

```python
dipu = df.groupby('Order_Date')['Total_Amount'].sum()

plt.figure(figsize=(20,5))

plt.plot(dipu.index, dipu.values)
plt.xticks(rotation=45)

plt.show()
```

24. Create a histogram of product prices.

```python
plt.figure(figsize=(7,3))


plt.hist(
    df['Price'],
    bins=5,
    color='green',
    edgecolor='orange'
)

plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Histogram of Product Prices')


plt.show()
```

25. Create a scatter plot of Quantity vs Total_Amount.

```python
plt.figure(figsize=(7,4))
plt.scatter(
    df['Quantity'],
    df['Total_Amount'],
    color='blue', 
    marker='o',
    label="project iteams"
)
plt.title("This is a scatter plot")
plt.xlabel("Quantity")
plt.ylabel("Total ammount")

plt.tight_layout()

plt.show()
```

26. Create a bar chart showing number of orders by city.

28. Create a stacked bar chart for Category and Payment Method.

```python
stack_data = df.groupby(['Category', 'Payment_Method'])['Total_Amount'].sum().unstack()

stack_data.plot(
    kind='bar',
    stacked=True,
    figsize=(8,6)
)

plt.title('Category and Payment Method')
plt.xlabel('Category')
plt.ylabel('Total Amount')

plt.tight_layout()

plt.show()
```

29. Create a horizontal bar chart for top 10 expensive products.

```python
print(df.columns)

expensive = df.sort_values(by='Price', ascending=False).head(10)

plt.barh(
    expensive["Product_Name"],
    expensive['Price']
)

plt.show()

```

30. Create a line chart showing monthly sales growth.

```python
print(df.columns)


# Convert Order_Date to datetime format
df['Order_Date'] = pd.to_datetime(df['Order_Date'])

# Create monthly sales data
monthly_sales = df.groupby(
    df['Order_Date'].dt.month
)['Total_Amount'].sum()

# Create figure
plt.figure(figsize=(8,5))

# Line chart
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker='o',
    color='blue'
)

# Labels and title
plt.title('Monthly Sales Growth')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.tight_layout()

plt.show()
```
