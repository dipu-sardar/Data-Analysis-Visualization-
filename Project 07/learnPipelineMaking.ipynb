•	ক্লাসের নাম দিন: EVBasicSummary
•	কনস্ট্রাক্টরে (__init__) শুধু file_path এবং _df থাকবে।
•	মেথড থাকবে মাত্র একটি: show_shape(), যা ডাটা লোড করে তার টোটাল রো এবং কলামের সংখ্যা (df.shape) প্রিন্ট করবে।

```python
# পাইথনে ডাটা অ্যানালিসিসের জন্য প্যান্ডাস লাইব্রেরি ইমপোর্ট করছি
import pandas as pd

# ১. ক্লাসের নাম ডিক্লেয়ার করছি (EVBasicSummary)
class EVBasicSummary:
    
    # ২. কনস্ট্রাক্টর পদ্ধতি বা ম্যাজিক মেথড (__init__)
    # যখনই আমরা এই ক্লাসের একটি নতুন অবজেক্ট তৈরি করব, এই মেথডটি স্বয়ংক্রিয়ভাবে রান হবে।
    def __init__(self, file_path: str):
        
        # 'self.file_path' একটি অ্যাট্রিবিউট, যা ফাইলের নাম বা লোকেশনটি মেমোরিতে মনে রাখবে।
        self.file_path = file_path
        
        # 'self._df' দিয়ে একটি খালি ভেরিয়েবল রাখলাম, যেখানে পরবর্তীতে আমরা আসল ডাটাফ্রেমটি লোড করে রাখব।
        # নামের শুরুতে আন্ডারস্কোর (_) দেওয়া মানে এটি প্রটেক্টেড বা এনক্যাপসুলেটেড (Protected Attribute)।
        self._df = None

    # ৩. আমাদের একমাত্র কাজের মেথড: show_shape()
    # এই ফাংশনটির কাজ হলো ডাটা লোড করা এবং তার ডাইমেনশন বা শেপ প্রদর্শন করা।
    def show_shape(self) -> None:
        
        # প্যান্ডাস ব্যবহার করে নির্দিষ্ট ফাইল পাথ থেকে ডাটা রিড করছি এবং ক্লাসের 'self._df' ভেরিয়েবলে রাখছি।
        # 'self.' ব্যবহার করার কারণে এই ডাটাফ্রেমটি ক্লাসের যেকোনো ফাংশন থেকে অ্যাক্সেস করা যাবে।
        self._df = pd.read_csv(self.file_path)
        
        # 'self._df.shape' প্যান্ডাসের একটি অ্যাট্রিবিউট যা একটি টাপল (Tuple) রিটার্ন করে (মোট রো, মোট কলাম)
        # আমরা সেই টাপলের প্রথম মান [0] দিয়ে মোট সারি এবং দ্বিতীয় মান [1] দিয়ে মোট কলাম আলাদা ভেরিয়েবলে নিচ্ছি।
        total_rows = self._df.shape[0]
        total_columns = self._df.shape[1]
        
        # প্রিন্ট ফাংশন এবং এফ-স্ট্রিং (f-string) ব্যবহার করে আউটপুটটি স্ক্রিনে সুন্দর করে দেখাচ্ছি।
        print("--- ডাটাসেটের শেপ বা ডাইমেনশন রিপোর্ট ---")
        print(f"টোটাল রো বা সারির সংখ্যা: {total_rows}")
        print(f"টোটাল কলাম বা ফিচারের সংখ্যা: {total_columns}")
```

```python
# ১. ক্লাসের একটি অবজেক্ট বা ইনস্ট্যান্স তৈরি করছি এবং ফাইলের নামটি পাস করছি।
# জাভাতে আমরা লিখতাম: EVBasicSummary summary = new EVBasicSummary("ev_charging_sessions.csv");
# পাইথনে 'new' কি-ওয়ার্ড লাগে না, সরাসরি ক্লাসের নাম ধরে কল করলেই অবজেক্ট তৈরি হয়।
summary_object = EVBasicSummary(file_path='ev_charging_sessions.csv')

# ২. এবার আমাদের অবজেক্টের ভেতর থেকে 'show_shape()' মেথড বা ফাংশনটিকে কল করছি।
summary_object.show_shape()
```
