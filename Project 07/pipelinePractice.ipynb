```python
#First of all I need to import all the libraries for data pipeline 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Now this is the time for making data pipeline with using Object Oriented Programming(OOP)

class EVBasicSummary:
    def __init__(self, file_path: str):
        self.file_path = file_path
        self._df = None
    
    def show_shape(self) -> None:
        self._df = pd.read_csv(self.file_path)

        total_row = self._df.shape[0]
        total_column = self._df.shape[1]

        print('====The dimention of Dataset====')
        print(f'Total Rows: {total_row}')
        print(f'Total Columns: {total_column}')
```
